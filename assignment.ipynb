```python
def subsets(arr):
    result = []

    def backtrack(i, current):
        if i == len(arr):
            result.append(current[:])
            return

        # Exclude arr[i]
        backtrack(i + 1, current)

        # Include arr[i]
        current.append(arr[i])
        backtrack(i + 1, current)
        current.pop()

    backtrack(0, [])
    return result

print(subsets([1, 2, 3]))
```

```python
def permutations(arr):
    result = []

    def backtrack(start):
        if start == len(arr):
            result.append(arr[:])
            return

        for i in range(start, len(arr)):
            arr[start], arr[i] = arr[i], arr[start]
            backtrack(start + 1)
            arr[start], arr[i] = arr[i], arr[start]

    backtrack(0)
    return result

print(permutations([1, 2, 3]))
```
